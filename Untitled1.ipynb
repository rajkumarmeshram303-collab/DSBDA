```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from scipy.stats import skew
```

```python
data = pd.read_csv("/content/sample_data/Salary_Data.csv")
```

```python
data.info()
```

```python
data.describe()
```

```python
mean = data['Salary'].mean()
median = data['Salary'].median()
mode = data['Salary'].mode()
sd = data['Salary'].std()

min_val = data['Salary'].min()
max_val = data['Salary'].max()

print("Mean =", mean)
print("Median =", median)
print("Mode =", mode.values)
print("Min =", min_val)
print("Max =", max_val)
print("Std Dev =", sd)
```

```python
plt.scatter(data['Age'], data['Salary'])
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary")
plt.show()
```

```python
(data['Salary'] > 50000).sum()
```

```python
skewness = (3 * (mean - median)) / sd
print(f"Skewness = {skewness}")
```

```python
plt.figure(figsize=(8, 5))

sns.kdeplot(data['Salary'])

plt.axvline(mean, label="Mean")
plt.axvline(median, color="black", label="Median")
plt.axvline(mode.iloc[0], color="green", label="Mode")

plt.title(f"Salary Distribution (Skewness = {skewness:.3f})")
plt.xlabel("Salary")
plt.legend()
plt.show()
```
