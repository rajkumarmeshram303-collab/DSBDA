```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data = pd.read_excel("/content/sample_data/bostonhousing.xlsx")
```

```python
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data = data.fillna(data.median(numeric_only=True))
data.isnull().sum()
```

```python
data.columns
```

```python
target = "medv"

X = data.drop(columns=[target])
y = data[target]
```

```python
X.isnull().sum()   # check
X = X.fillna(X.median(numeric_only=True))
```

```python
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
from sklearn.linear_model import LinearRegression

lm = LinearRegression()
lm.fit(xtrain, ytrain)
```

```python
ytrain_pred = lm.predict(xtrain)
ytest_pred = lm.predict(xtest)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

mean_squared_error(ytrain, ytrain_pred)
mean_squared_error(ytest, ytest_pred)

r2_score(ytrain, ytrain_pred)
r2_score(ytest, ytest_pred)
```

```python
train_result = pd.DataFrame({"Actual": ytrain.values, "Predicted": ytrain_pred})
test_result  = pd.DataFrame({"Actual": ytest.values,  "Predicted": ytest_pred})

train_result.head()
test_result.head()
```

```python

plt.scatter(ytrain, ytrain_pred, c='blue', marker='o', label='Training data')
plt.scatter(ytest, ytest_pred, c='lightgreen', marker='s', label='Test data')

plt.xlabel('True values')
plt.ylabel('Predicted values')
plt.title("True value vs Predicted value")
plt.legend(loc='upper left')
plt.show()
```
